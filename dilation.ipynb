```python
# Import packages
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("wallp.png")

# Dilation sizes
d1 = 0
d2 = 10
d3 = 20

# Dilation types
t1 = cv2.MORPH_RECT      # Rectangle
t2 = cv2.MORPH_CROSS     # Cross
t3 = cv2.MORPH_ELLIPSE   # Ellipse

# Create dilation kernels
tmp1 = cv2.getStructuringElement(
    t1,
    (2*d1 + 1, 2*d1 + 1),
    (d1, d1)
)

tmp2 = cv2.getStructuringElement(
    t2,
    (2*d2 + 1, 2*d2 + 1),
    (d2, d2)
)

tmp3 = cv2.getStructuringElement(
    t3,
    (2*d3 + 1, 2*d3 + 1),
    (d3, d3)
)

# Apply dilation
final1 = cv2.dilate(image, tmp1)
final2 = cv2.dilate(image, tmp2)
final3 = cv2.dilate(image, tmp3)

# Convert BGR to RGB for display
final1 = cv2.cvtColor(final1, cv2.COLOR_BGR2RGB)
final2 = cv2.cvtColor(final2, cv2.COLOR_BGR2RGB)
final3 = cv2.cvtColor(final3, cv2.COLOR_BGR2RGB)

# Display results
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(final1)
plt.title("Rectangle Dilation")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(final2)
plt.title("Cross Dilation")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(final3)
plt.title("Ellipse Dilation")
plt.axis("off")

plt.show()
```
