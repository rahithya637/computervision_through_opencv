```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("palm.jpeg")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect palm lines
edges = cv2.Canny(gray, 60, 150)

# Convert edges to 3-channel image
edges_color = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
plt.figure(figsize=(8,6))
plt.imshow(edges_color, cmap='gray')
plt.title("Palm Lines Detection")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("palm.jpeg")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect palm lines
edges = cv2.Canny(gray, 50, 150)

# Convert edges to 3-channel image
edges_color = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)

# Blend original image and detected lines
blended = cv2.addWeighted(
    img,          # Original image
    0.5,          # Weight of original image
    edges_color,  # Edge image
    0.5,          # Weight of edge image
    0
)

# Display result
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(blended, cv2.COLOR_BGR2RGB))
plt.title("Blended Output")
plt.axis("off")

plt.show()
```
