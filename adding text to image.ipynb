```python
# Import packages
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("wallp.png")

# Add text to image
cv2.putText(
    image,
    "OpenCV",                  # Text
    (50, 100),                 # Position (x,y)
    cv2.FONT_HERSHEY_SIMPLEX,  # Font
    2,                         # Font Scale
    (0, 255, 0),               # Color (Green)
    3                          # Thickness
)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display image
plt.imshow(image)
plt.title("Image with Text")
plt.axis("off")
plt.show()
```
