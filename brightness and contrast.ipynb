```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("wallp.png")

# Increase brightness
bright = cv2.convertScaleAbs(img, alpha=1, beta=50)

plt.imshow(cv2.cvtColor(bright, cv2.COLOR_BGR2RGB))
plt.title("Brightness Increased")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("wallp.png")

# alpha > 1 increases contrast
contrast = cv2.convertScaleAbs(img, alpha=2, beta=0)

plt.imshow(cv2.cvtColor(contrast, cv2.COLOR_BGR2RGB))
plt.title("Contrast Increased")
plt.axis("off")
plt.show()
```
